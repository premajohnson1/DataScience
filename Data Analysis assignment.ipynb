# 7)Test the Analysis of Variance between etest_p and mba_p at signifance level 5%.(Make decision using Hypothesis Testing)

```python
Null Hypothesis (H₀): There is no significant difference between the mean of etest_p and mba_p.
Alternative Hypothesis (H₁): There is a significant difference between the mean of etest_p and mba_p.
```

```python
import scipy.stats as stats

stats.f_oneway(dataset['etest_p'], dataset['mba_p'])
```

```python
Here,
- P-value = 0.0000000000000000000467 (approximately)
- Significance level = 0.05
Since P-value < 0.05, we reject the null hypothesis.
```

```python
There is a statistically significant difference between the mean of Employability Test Percentage (etest_p) and MBA Percentage (mba_p) at the 5% significance level.
```

# 8)Test the similarity between the degree_t(Sci&Tech) and specialisation(Mkt&HR) with respect to salary at significance level of 5%.(Make decision using Hypothesis Testing)

```python
placed = dataset[dataset['status'] == 'Placed']

group1 = placed[placed['degree_t'] == 'Sci&Tech']['salary']

group2 = placed[placed['specialisation'] == 'Mkt&HR']['salary']
```

```python
Null Hypothesis (H₀): There is no significant difference in the average salary of Sci&Tech degree students and Mkt&HR specialization students.
Alternative Hypothesis (H₁): There is a significant difference in their average salaries.
```

```python
import scipy.stats as stats

stats.ttest_ind(group1, group2, equal_var=False)
```

```python
Here:
- P-value = 0.01341
- Significance level = 0.05
Since 0.01341 < 0.05, we reject the null hypothesis.

There is a statistically significant difference between the average salaries of Sci&Tech degree students and Mkt&HR specialization students at the 5% significance level.
```

# 9)Convert the normal distribution to standard normal distribution for salary column

```python
To convert the Salary column into a Standard Normal Distribution, we use Standardization (Z-score normalization).
Step 1: Understand Standardization
Standardization converts the original salary values into Z-scores.

The resulting values have:
- Mean = 0
- Standard Deviation = 1
```

```python
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()

dataset['salary'] = sc.fit_transform(dataset[['salary']])
```

# 10)What is the probability Density Function of the salary range from 700000 to 900000?

# 11)Test the similarity between the degree_t(Sci&Tech)with respect to etest_p and mba_p at significance level of 5%.(Make decision using Hypothesis Testing)

```python
sci = dataset[dataset['degree_t'] == 'Sci&Tech']
```

```python
Null Hypothesis (H₀): There is no significant difference between the mean etest_p and mba_p of Sci&Tech students.
Alternative Hypothesis (H₁): There is a significant difference between the mean etest_p and mba_p of Sci&Tech students.
```

```python
import scipy.stats as stats

stats.ttest_rel(sci['etest_p'], sci['mba_p'])
```

```python
Here:
- P-value = 0.0000055179
- Significance level = 0.05
Since P-value < 0.05, we reject the null hypothesis.
```

```python
print(sci['etest_p'].mean())
print(sci['mba_p'].mean())
```

```python
There is a statistically significant difference between the average Employability Test Percentage and MBA Percentage among Sci&Tech students at the 5% significance level.
```
