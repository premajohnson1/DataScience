```python
import pandas as pd
from statsmodels.stats.outliers_influence import variance_inflation_factor

def calc_vif(X):

    # Calculating VIF
    vif = pd.DataFrame()

    vif["variables"] = X.columns

    vif["VIF"] = [
        variance_inflation_factor(X.values, i)
        for i in range(X.shape[1])
    ]

    return vif


# Calling the function
calc_vif(dataset[["ssc_p","hsc_p","degree_p","etest_p","mba_p"]])

```

# Homoscedasticity = Constant variance.

# Heteroscedasticity = Unequal variance.

# Residual plot = Helps identify the pattern.

# Breusch-Pagan test = Statistical test for heteroscedasticity.

```python
# These two concepts are especially important in Linear Regression.

# Residual = Actual value − Predicted value

# Example:
# Actual salary	Predicted salary	Residual
# 25,000	24,000	1,000
# 30,000	28,000	2,000
# 40,000	42,000	-2,000
# 50,000	46,000	4,000

# Residual means the difference between the actual output and the predicted output.

# 2. What is Homoscedasticity?
# Definition: Homoscedasticity means the variance (spread) of residuals remains approximately constant across different predicted values or levels of the independent variable.

# In simple English:

# The prediction errors have approximately the same spread throughout the dataset.

# Example:

# Imagine predicting salaries:
# - For a predicted salary of ₹20,000, errors are around ₹2,000.
# - For a predicted salary of ₹30,000, errors are around ₹2,000.
# - For a predicted salary of ₹40,000, errors are around ₹2,000.
# The error spread remains approximately constant.

# This is called Homoscedasticity.
# 3. What is Heteroscedasticity?
# Definition: Heteroscedasticity means the variance (spread) of residuals changes across different predicted values or levels of the independent variable.

# In simple English:

# The prediction errors have different spreads in different parts of the dataset.

# Example:

# Imagine predicting salaries:
# - For a predicted salary of ₹20,000, errors are around ₹1,000.
# - For a predicted salary of ₹30,000, errors are around ₹4,000.
# - For a predicted salary of ₹40,000, errors are around ₹8,000.
# The error spread increases as the predicted salary increases.

# This is called Heteroscedasticity.
# 4. Difference between Homoscedasticity and Heteroscedasticity
# Homoscedasticity	Heteroscedasticity
# Constant variance	Changing variance
# Residual spread is similar	Residual spread is different
# Error variance remains stable	Error variance increases or decreases
# Desirable for classical Linear Regression assumptions	Indicates a violation of the constant variance assumption

# 5. Which one is best for a Machine Learning model?
# Homoscedasticity is preferred
# For classical Linear Regression, homoscedasticity is preferred because:
# 1. Error variance remains constant.
# 2. Regression coefficient estimates have better statistical efficiency under the usual assumptions.
# 3. Standard statistical tests and confidence intervals are more reliable.
# However, remember:

# Heteroscedasticity does not mean your ML model is useless or its predictions are necessarily inaccurate.

# It mainly affects statistical inference, such as standard errors, confidence intervals and hypothesis tests.

# Also, homoscedasticity is not a compulsory assumption for every Machine Learning algorithm.
```

```python
# Imagine a teacher wants to know whether two teaching methods produce different student marks.
# - Method A: Traditional teaching
# - Method B: Online teaching
# The teacher collects marks from students who learned using both methods.

# Now the teacher wants to check:

# Are the average marks of these two groups different?

# For this, we use Hypothesis Testing.
# 2. Types of Hypothesis
# There are two main types:
# A. Null Hypothesis (H₀)
# Null Hypothesis means there is no statistically significant difference or relationship (depending on what we are testing).

# Example:

# The average marks of students taught using Method A and Method B are equal.

# B. Alternative Hypothesis (H₁)
# Alternative Hypothesis means there is a statistically significant difference or relationship.

# Example:

# The average marks of students taught using Method A and Method B are different.
# --------------------

# Understand this with a real-life analogy
# Imagine a company claims:

# "Our new training programme improves employee performance."

# We collect employee performance scores before and after the training.

# Our hypotheses could be:
# - H₀: The training programme does not change the average performance score.
# - H₁: The training programme changes the average performance score.
# We perform a paired T-test because we are comparing the same employees before and after training.

# If p-value is 0.02:
# - 0.02 is less than 0.05.
# - Reject H₀.
# - There is evidence that the average performance score changed after training.
# Notice: If the alternative hypothesis specifically says improves, we should use a one-tailed test and define the direction before testing.


# --------------

# 6. Important terms to remember
# Term	Simple meaning
# Hypothesis	An assumption we want to test
# H₀	Null hypothesis
# H₁	Alternative hypothesis
# α (Alpha)	Significance level, commonly 0.05
# P-value	Probability of results this extreme or more extreme, assuming H₀ is true
# Reject H₀	Evidence against the null hypothesis
# Fail to reject H₀	Insufficient evidence against the null hypothesis

```

# ANOVA problem statements for practicing both One-Way ANOVA and Two-Way ANOVA in Machine Learning / Statistics.

1. One-Way ANOVA – Problem Statements
Meaning: One-Way ANOVA is used to compare the means of a numerical variable across three or more independent groups based on one categorical factor.

Problem 1: Student Performance
- Numerical variable: Placement salary
- Categorical variable: Degree (Sci&Tech, Comm&Mgmt, Others)
- Question: Is there a significant difference in average placement salary among students from different degree groups?
  
Problem 2: Teaching Methods
- Numerical variable: Exam marks
- Categorical variable: Teaching method (Online, Offline, Hybrid)
- Question: Does the teaching method significantly affect students' average exam marks?

Problem 3: Employee Salary
- Numerical variable: Salary
- Categorical variable: Department (IT, HR, Sales, Finance)
- Question: Is there a significant difference in average salaries among different departments?
  
Problem 4: Customer Satisfaction
- Numerical variable: Satisfaction score
- Categorical variable: Service type (Basic, Standard, Premium)
- Question: Does the service type significantly influence customer satisfaction scores?
  
2. Two-Way ANOVA – Problem Statements
Meaning: Two-Way ANOVA is used to study the effect of two categorical independent variables on one numerical dependent variable. It can also examine their interaction effect.

Problem 1: Student Performance
- Numerical variable: Exam marks
- Categorical variable 1: Teaching method (Online, Offline)
- Categorical variable 2: Gender (Male, Female)
- Question:
  1. Does teaching method affect exam marks?
  2. Does gender affect exam marks?
  3. Is there an interaction effect between teaching method and gender?
     
Problem 2: Employee Salary
- Numerical variable: Salary
- Categorical variable 1: Department (IT, HR, Sales)
- Categorical variable 2: Experience level (Fresher, Experienced)
- Question: Does department, experience level, or their interaction significantly affect employee salary?
  
Problem 3: Crop Production
- Numerical variable: Crop yield
- Categorical variable 1: Fertilizer type (A, B, C)
- Categorical variable 2: Irrigation method (Drip, Sprinkler)
- Question: Do fertilizer type, irrigation method, or their interaction significantly affect crop yield?
  
Problem 4: Marketing Analysis
- Numerical variable: Sales revenue
- Categorical variable 1: Advertisement type (TV, Social Media, Newspaper)
- Categorical variable 2: Region (North, South, East)
- Question: Do advertisement type, region, or their interaction significantly affect sales revenue?
